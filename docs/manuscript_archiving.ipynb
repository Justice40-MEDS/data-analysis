# Manuscript Data Wrangling for Archival

```python
# Load packages 
import pandas as pd
import geopandas as gpd
```

```python
pd.set_option('display.max_columns', None)

# Read in GeoJSON of all 
burdens = gpd.read_file('/home/kat/github/data-analysis/data/december-geoda-output-burdens.geojson')

indicators = gpd.read_file('/home/kat/github/data-analysis/data/december-geoda-output-burdens.geojson')

```

```python
indicators.head()
```

```python
# Wrangling burdens geoda data
burdens_cleaned = burdens[['GEOID10', 'SF', 'CF', 'CC', 'TC', 'G_STR', 'C_ID', 'PP_VAL', 'geometry']].rename(columns={
    'SF': 'State',
    'CF': 'County',
    'CC': 'Total_Burdens',
    'TC': 'Total_Indicators',
    'C_ID': 'C_ID_Burdens',
    'G_STR': 'Gi_Z_Burdens',
    'PP_VAL': 'Gi_P_Burdens'
})
burdens_cleaned.head()
```

```python
# Wrangling burdens geoda data
indicators_cleaned = indicators[['GEOID10', 'SF', 'CF', 'CC', 'TC', 'G_STR', 'C_ID', 'PP_VAL', 'geometry']].rename(columns={
    'SF': 'State',
    'CF': 'County',
    'CC': 'Total_Burdens',
    'TC': 'Total_Indicators',
    'C_ID': 'C_ID_Indicators',
    'G_STR': 'Gi_Z_Indicators',
    'PP_VAL': 'Gi_P_Indicators'
})
indicators_cleaned.head()
```

```python
# Check if the three columns match between dataframes
match_check = burdens_cleaned[['GEOID10', 'Total_Burdens', 'Total_Indicators']].equals(
    indicators_cleaned[['GEOID10', 'Total_Burdens', 'Total_Indicators']]
)

print(match_check)
```

```python
# Join data on overlapping columns
overlap = set(burdens_cleaned.columns) & set(indicators_cleaned.columns)

joined = burdens_cleaned.merge(
    indicators_cleaned,
    on=list(overlap),
    how='inner'
)

joined.head()
```

```python
# Burdens
joined['VCS_burdens'] = ((joined['C_ID_Burdens'] == 2) & (joined['Gi_P_Burdens'] < 0.01)).astype(int)
joined['CS_burdens'] = ((joined['C_ID_Burdens'] == 2) & (joined['Gi_P_Burdens'] >= 0.01) & (joined['Gi_P_Burdens'] < 0.05)).astype(int)
joined['HS_burdens'] = ((joined['C_ID_Burdens'] == 1) & (joined['Gi_P_Burdens'] >= 0.01) & (joined['Gi_P_Burdens'] < 0.05)).astype(int)
joined['VHS_burdens'] = ((joined['C_ID_Burdens'] == 1) & (joined['Gi_P_Burdens'] < 0.01)).astype(int)

# Indicators
joined['VCS_indicators'] = ((joined['C_ID_Indicators'] == 2) & (joined['Gi_P_Indicators'] < 0.01)).astype(int)
joined['CS_indicators'] = ((joined['C_ID_Indicators'] == 2) & (joined['Gi_P_Indicators'] >= 0.01) & (joined['Gi_P_Indicators'] < 0.05)).astype(int)
joined['HS_indicators'] = ((joined['C_ID_Indicators'] == 1) & (joined['Gi_P_Indicators'] >= 0.01) & (joined['Gi_P_Indicators'] < 0.05)).astype(int)
joined['VHS_indicators'] = ((joined['C_ID_Indicators'] == 1) & (joined['Gi_P_Indicators'] < 0.01)).astype(int)

# Drop C_ID columns
# joined = joined.drop(columns=['C_ID_Burdens', 'C_ID_Indicators'])

joined = joined[[c for c in joined.columns if c != 'geometry'] + ['geometry']]
```
